```python
# ============================================================
# BANKRUPTCY PREDICTION - COMPLETE EDA
# Dataset: Taiwanese Bankruptcy Prediction
# ============================================================

# -------------------------------
# 1. IMPORT LIBRARIES
# -------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import os

from google.colab import files

print("Libraries imported successfully!")


# -------------------------------
# 2. UPLOAD ZIP FILE
# -------------------------------

print("Please upload the Taiwanese Bankruptcy Prediction ZIP file.")

uploaded = files.upload()

# Check if any file was uploaded
if not uploaded:
    raise FileNotFoundError("No file was uploaded. Please make sure to upload the ZIP file.")

zip_file = list(uploaded.keys())[0]

print("\nUploaded file:", zip_file)


# -------------------------------
# 3. EXTRACT ZIP FILE
# -------------------------------

extract_folder = "bankruptcy_dataset"

with zipfile.ZipFile(zip_file, "r") as zip_ref:
    zip_ref.extractall(extract_folder)

print("\nZIP file extracted successfully!")


# -------------------------------
# 4. FIND CSV FILE
# -------------------------------

csv_file = None

for root, dirs, files_list in os.walk(extract_folder):
    for file in files_list:
        if file.endswith(".csv"):
            csv_file = os.path.join(root, file)
            break

    if csv_file:
        break

if csv_file is None:
    raise FileNotFoundError("No CSV file found inside the ZIP file.")

print("CSV file found:", csv_file)


# -------------------------------
# 5. LOAD DATASET
# -------------------------------

df = pd.read_csv(csv_file)

print("\nDataset loaded successfully!")


# ============================================================
#                    EDA STARTS HERE
# ============================================================


# -------------------------------
# 6. FIRST FIVE ROWS
# -------------------------------

print("\n" + "="*60)
print("1. FIRST FIVE ROWS")
print("="*60)

display(df.head())


# -------------------------------
# 7. DATASET SHAPE
# -------------------------------

print("\n" + "="*60)
print("2. DATASET SHAPE")
print("="*60)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# -------------------------------
# 8. COLUMN NAMES
# -------------------------------

print("\n" + "="*60)
print("3. COLUMN NAMES")
print("="*60)

print(df.columns.tolist())


# -------------------------------
# 9. DATA TYPES
# -------------------------------

print("\n" + "="*60)
print("4. DATA TYPES")
print("="*60)

print(df.dtypes)


# -------------------------------
# 10. MISSING VALUES
# -------------------------------

print("\n" + "="*60)
print("5. MISSING VALUES")
print("="*60)

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal missing values:",
      df.isnull().sum().sum())


# -------------------------------
# 11. DUPLICATE ROWS
# -------------------------------

print("\n" + "="*60)
print("6. DUPLICATE ROWS")
print("="*60)

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# -------------------------------
# 12. DESCRIPTIVE STATISTICS
# -------------------------------

print("\n" + "="*60)
print("7. DESCRIPTIVE STATISTICS")
print("="*60)

display(df.describe())


# -------------------------------
# 13. TARGET VARIABLE
# -------------------------------

target = "Bankrupt?"

print("\n" + "="*60)
print("8. TARGET VARIABLE DISTRIBUTION")
print("="*60)

target_counts = df[target].value_counts()

print(target_counts)

print("\nPercentage distribution:")

target_percentage = (
    df[target]
    .value_counts(normalize=True)
    * 100
)

print(target_percentage)


# -------------------------------
# 14. CLASS DISTRIBUTION GRAPH
# -------------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Bankruptcy Class Distribution")
plt.xlabel("Bankrupt? (0 = Non-Bankrupt, 1 = Bankrupt)")
plt.ylabel("Number of Companies")

plt.show()


# -------------------------------
# 15. CONSTANT FEATURES
# -------------------------------

print("\n" + "="*60)
print("9. CONSTANT FEATURES")
print("="*60)

constant_features = []

for column in df.columns:

    if df[column].nunique() <= 1:
        constant_features.append(column)

print("Constant features:")

if len(constant_features) == 0:
    print("No constant features found.")
else:
    print(constant_features)


# -------------------------------
# 16. NUMERICAL FEATURES
# -------------------------------

print("\n" + "="*60)
print("10. NUMERICAL FEATURES")
print("="*60)

numeric_df = df.select_dtypes(include=np.number)

print("Number of numerical columns:",
      numeric_df.shape[1])


# -------------------------------
# 17. REPLACE INFINITY VALUES
# -------------------------------

df_clean = df.replace(
    [np.inf, -np.inf],
    np.nan
)

print("\nInfinity values replaced.")


# -------------------------------
# 18. FEATURE-TARGET CORRELATION
# -------------------------------

print("\n" + "="*60)
print("11. FEATURE-TARGET CORRELATION")
print("="*60)

numeric_df = df_clean.select_dtypes(
    include=np.number
)

correlation = numeric_df.corr()[target]

correlation = correlation.drop(target)

correlation = correlation.abs().sort_values(
    ascending=False
)

print("\nTop 15 features correlated with bankruptcy:")

display(correlation.head(15))


# -------------------------------
# 19. CORRELATION HEATMAP
# -------------------------------

top_features = correlation.head(10).index.tolist()

heatmap_data = df_clean[
    top_features + [target]
].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Correlation Heatmap - Top Financial Features"
)

plt.show()


# -------------------------------
# 20. BOXPLOTS
# -------------------------------

print("\n" + "="*60)
print("12. BOXPLOTS")
print("="*60)

for feature in top_features[:4]:

    plt.figure(figsize=(7,5))

    sns.boxplot(
        data=df_clean,
        x=target,
        y=feature
    )

    plt.title(
        f"{feature} vs Bankruptcy"
    )

    plt.xlabel(
        "Bankrupt? (0 = Non-Bankrupt, 1 = Bankrupt)"
    )

    plt.ylabel(feature)

    plt.show()


# -------------------------------
# 21. EDA SUMMARY
# -------------------------------

print("\n" + "="*60)
print("EDA SUMMARY")
print("="*60)

print("Dataset size:", df.shape)

print(
    "Total missing values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Number of numerical features:",
    numeric_df.shape[1]
)

print(
    "Non-Bankrupt companies:",
    (df[target] == 0).sum()
)

print(
    "Bankrupt companies:",
    (df[target] == 1).sum()
)

print("\nTop 5 correlated features:")

display(correlation.head(5))

print("\nEDA COMPLETED SUCCESSFULLY!")
```
